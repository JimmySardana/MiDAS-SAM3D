```python
from pathlib import Path
import requests

SERVER_URL = "http://localhost:8001"   # Change port if needed
IMAGE_PATH = "images/image.png"
MASK_PATH  = "images/mask.png"
OUTPUT_PATH = "output/result.glb"
```

```python
resp = requests.get(f"{SERVER_URL}/health", timeout=10)
print(resp.status_code, resp.json())
```

```python
IMAGE_PATH = "images/image.png"
MASK_PATH = "images/mask.png"


# -----------------------
# SEND REQUEST
# -----------------------
with open(IMAGE_PATH, "rb") as image, open(MASK_PATH, "rb") as mask:
    files = {
        "image": ("image.png", image, "image/png"),
        "mask": ("mask.png", mask, "image/png")
    }

    print("[CLIENT] Sending request...")
    resp = requests.post(f"{SERVER_URL}/infer/glb", files=files, timeout=300)

# ---- Check response ----
print(f"[CLIENT] Status : {resp.status_code}")
print(f"[CLIENT] Content-Type: {resp.headers.get('Content-Type')}")

# ---- Save GLB bytes directly ----
Path(OUTPUT_PATH).write_bytes(resp.content)
print(f"[CLIENT] Saved {len(resp.content) / 1024:.1f} KB -> {OUTPUT_PATH}")
```
